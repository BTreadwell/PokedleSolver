```python
types = ["none", "normal", "fighting", "flying", "poison", "ground", "rock", "bug","ghost","steel","fire","water","grass","electric","psychic","ice", "dragon","dark","fairy"
]

colors = ["none","black","blue","brown","gray","green","pink", "purple", "red", "white", "yellow"]
```

```python
with open("Data/types.csv", 'w') as f:
    f.write(",".join(types) + "\n")

with open("Data/colors.csv", 'w') as f:
    f.write(",".join(colors) + "\n")
```

```python
names = ['none']
with open("pokemon_species.csv", 'r') as f:
    f.readline()
    for line in f:
        name = line.strip().split(',')[1]
        names.append(name)

with open('Data/names.csv', 'w') as f:
    f.write(",".join(names) + "\n")
```

```python
from collections import defaultdict

scrapped_data = {}
with open("pokemon_species.csv", 'r') as f:
    f.readline()
    for line in f:
        data = line.strip().split(',')
        scrapped_data[int(data[0])] = [int(data[2]), data[3], int(data[4]), int(data[5])]

types = defaultdict(lambda: [0, 0])
with open("pokemon_types.csv", 'r') as f:
    f.readline()
    for line in f:
        data = line.strip().split(',')
        types[int(data[0])][int(data[2]) - 1] = int(data[1])

stages = [0] * 1026
def get_stage(id):
    if stages[id] != 0:
        return stages[id]
    elif scrapped_data[id][1] == "":
        return 1
    else:
        return get_stage(int(scrapped_data[id][1])) + 1

for id in scrapped_data.keys():
    if stages[id] == 0:
        stages[id] = get_stage(id)

is_final = [1] * 1026
for id in scrapped_data.keys():
    if scrapped_data[id][1] != "":
        is_final[int(scrapped_data[id][1])] = 0


pokemon_data = []

for id in scrapped_data.keys():
    p = scrapped_data[id]
    gen = p[0]
    t1 = types[id][0]
    t2 = types[id][1]
    stage = stages[id]
    final_evo = is_final[id]
    color = p[3]

    pokemon_data.append([id, gen, t1, t2, stage, final_evo, color])


```

```python
with open('Data/pokemon.csv', 'w') as f:
    for x in pokemon_data:
        f.write(",".join([str(y) for y in x]) + "\n")
```
